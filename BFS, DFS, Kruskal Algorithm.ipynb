```python
from collections import deque
```

# BFS, DFS, Kruskal Algorithm

## Adjazenzliste

```python
def create_adjacentlist (edges):
    adj = {}
    for u, v, w in edges:
        if u not in adj:
            adj[u] = {}
        if v not in adj:
            adj[v] = {}

        if v not in adj[u]:
            adj[u][v] = w
            adj[v][u] = w
        else:
            adj[u][v] = min (adj[u][v], w)
            adj[v][u] = min (adj[u][v], w)

    return adj
```

## Initialisation

```python
def initialize (edges):
    nodes = set()
    for u, v, _ in edges:
        nodes.add(u)
        nodes.add(v)
    nodes = list(nodes)

    idx = {node: i for i, node in enumerate(nodes)}

    return idx, nodes
```

## BFS

```python
def bfs (edges, start):
    idx, nodes = initialize (edges)
    adj = create_adjacentlist (edges)

    pred = [None] * len(nodes)
    pred[idx[start]] = start

    q = deque()
    q.append(start)

    visited = [False] * len(nodes)
    visited[idx[start]] = True

    while q:
        u = q.popleft()
        for v in adj[u]:
            if not visited[idx[v]]:
                pred[idx[v]] = u
                visited[idx[v]] = True
                q.append(v)

    return pred
```

## DFS

```python
def dfs (edges, start):
    idx, nodes = initialize (edges)
    adj = create_adjacentlist (edges)

    visited = [False] * len(nodes)

    pred = [None] * len(nodes)
    pred[idx[start]] = start

    stack = [(start, start)]

    while stack:
        u, parent = stack.pop()
        if not visited[idx[u]]:
            visited[idx[u]] = True
            pred[idx[u]] = parent

            for v in adj[u]:
				if not visited[idx[v]]:
					stack.append((v,u))

    return pred
```

## Kruskal Algorithm

```python
def kruskal (edges):
    idx, nodes = initialize (edges)

    mst_edges = []
    total_weight = 0

    sorted_edges = sorted(edges, key = lambda x: x[2])

    for u, v, w in sorted_edges:
        if not mst_edges:
            mst_edges.append((u, v))
            total_weight += w

        pred = bfs (mst_edges, u)

        if pred[idx[v]] is None:
            mst_edges.append((u, v))
            total_weight += w

            if len(mst_edges) == len(nodes)-1:
                break

    return mst_edges, total_weight
```
